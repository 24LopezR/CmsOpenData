# How to do an Analysis in Particle Physics?

## Introduction

The aim of the exercise is to identify muons in the CMS detector at the LHC using the Z boson decay, by scanning data events and classify them as Z → μμ. The work will be distributed in three different exercises following sequentially the different steps of the proposed analysis. The main goals of this practical exercise are:

* Learn the workflow of a particle physics data analysis.
* Discriminate Z → µµ events from other backgrounds with 2µ in the final state.
* Measure the mass and width of the Z boson.
* Understand the importance of experiments in Particle Physics.

"RDFAnalyzer" is a set of codes keeping all necessary tools needed to carry out a basic analysis in Particle Physics. It's been developed to be used by junior and senior students in the Physics Bachelor. The data we are going to analyze is a small subset of the data recorded in CMS in 2016, at a center-of-mass energy of 13 TeV.

## Work Tools 

The package software uses interactively the ROOT program (version 6.24/06), dedicated to large data analysis, with a Python interpreter (Python 3.9) due to a Python extension module [PyROOT](https://root.cern.ch/pyroot). 

The framework is running over docker containers running over the opendata platform at IFCA (https://opendata.ifca.es/).

The code exploits the capabilities of ROOT's RDataFrames. RDataFrame offers a modern, high-level interface for analysis of data stored in TTree , CSV and other data formats, in C++ or Python.

Further documentation on ROOT and RDataFrames can be found here:
* https://root.cern.ch/
* https://root.cern.ch/root/htmldoc/guides/users-guide/Trees.html
* https://root.cern.ch/root/htmldoc/guides/users-guide/Histograms.html
* https://root.cern/doc/v628/classROOT_1_1RDataFrame.html 

## Task Description 

Starting with all physical events in our dataset, you will be asked to classify the events following a physical selection criteria, deciding which muons could have been produced in a Z->mumu decay.

You already have some basics about how the detector works, e.g. what kind of 'traces' the different particles leave in the various detectors (tracker, calorimeters, muon chambers) and which are the main kinematics variables in a modern experimental particle physics standard analysis. 

The practical exercise will follow these steps: 

1. Load the data and understand ROOT's ntuple structure.

2. Muon identification and isolation: display different histograms for the main kinematics variables used in the muon identification and isolation. Study the different variables and establish the selection criteria.

3. Candidates selection: Apply the selection criteria from point 2 and study the kinematic variables of the prompt muons you have selected, in particular, the invariant mass. 
 
4. Analysis: Repeat steps 2 and 3 until an optimal selection for Z → µµ events is achieved. 

5. Z boson measurement: find out the invariant mass of opposite changed mu+ mu- pairs and identify the Z → μμ candidates. Do a statistical analysis of the results to measure the Z boson properties (Z mass and decay width).

## Practical exercise
 
#### [Part0](./notebooks/Part0_Load_dataset.ipynb): Load the dataset from CMSOpenData. 
* Load the dataset with ROOT
* Understand flat NTuple/NANOAOD format.

#### [Part1](./notebooks/Part1_Analysis_workflow.ipynb): Analysis workflow.
* Start the Analysis and create the first histograms
* Display the histograms and study the different variables to indentify and isolate prompt muons.
* Design a selection to discriminate Z → µµ events from other backgrounds.
* Apply the selection and visualize the distributions of the kinematic variables.
* Optimize the selection so that the Z boson clearly appears in the dimuon invariant mass distribution.

#### [Part2](./notebooks/Part2_Z_boson_fit.ipynb): Measurement of Z boson mass and width. 

* Fit the dimuon invariant mass distribution to different functions and extract the mass and width of the Z boson.
* Analyze whic of the functions fits better the Z boson peak.
